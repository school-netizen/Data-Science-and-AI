## 1. Importing Required Libraries

The following libraries are used for data analysis and visualization:

- **Pandas** is used for loading and manipulating the dataset.
- **NumPy** provides numerical operations.
- **Matplotlib** is used for creating and customizing plots.
- **Seaborn** provides high-level statistical visualizations.

I also configure the plotting style and figure size for better readability.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["font.size"] = 11

print("Libraries imported successfully.")
```

## 2. Loading the Titanic Dataset

The Titanic dataset contains information about passengers such as age, gender, passenger class, fare, family relationships, and survival status.

I load the dataset into a Pandas DataFrame so that it can be analyzed and visualized.

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)

df.head()
```

## 3. Understanding the Dataset

Before creating visualizations, I inspect the dataset to understand its structure and identify the variables that can be visualized.

The dataset contains both numerical and categorical variables. Numerical variables such as Age and Fare are useful for histograms, box plots, and scatter plots, while categorical variables such as Sex and Pclass are useful for bar charts.

```python
df.info()
```

```python
df.describe()
```

## 4. Histogram: Age Distribution

A histogram shows how numerical values are distributed across different ranges.

Here, I use a histogram to examine the age distribution of Titanic passengers. The KDE curve provides an additional estimate of the underlying distribution.

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="Age",
    bins=30,
    kde=True
)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.show()
```

## 5. Histogram: Fare Distribution

The Fare variable represents the amount paid for a passenger's ticket.

A histogram helps identify the distribution of fares and whether the dataset contains unusually high or low fare values.

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="Fare",
    bins=30,
    kde=True
)

plt.title("Fare Distribution")
plt.xlabel("Ticket Fare")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.show()
```

## 6. Bar Chart: Passenger Count by Gender

Bar charts are useful for comparing different categorical groups.

In this visualization, I compare the number of male and female passengers in the dataset.

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Sex",
    hue="Sex",
    palette="Set2",
    legend=False
)

plt.title("Number of Passengers by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.show()
```

## 7. Bar Chart: Passengers by Class

The Titanic dataset contains three passenger classes.

This bar chart compares the number of passengers belonging to each class and helps visualize the distribution of passengers across different travel classes.

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Pclass",
    palette="viridis",
    legend=False
)

plt.title("Number of Passengers by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.show()
```

## 8. Bar Chart: Survival by Gender

This visualization compares survival outcomes between male and female passengers.

The `hue` parameter separates survivors and non-survivors, allowing the survival distribution within each gender group to be compared visually.

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Sex",
    hue="Survived",
    palette="Set1"
)

plt.title("Survival Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.legend(
    title="Survived",
    labels=["No", "Yes"]
)

plt.tight_layout()
plt.show()
```

## 9. Bar Chart: Survival by Passenger Class

This chart compares survival outcomes across first, second, and third class passengers.

It provides a visual way to examine whether survival counts differed across passenger classes.

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Survived",
    palette="Set1"
)

plt.title("Survival Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.legend(
    title="Survived",
    labels=["No", "Yes"]
)

plt.tight_layout()
plt.show()
```

## 10. Box Plot: Age by Survival Status

A box plot summarizes the distribution of numerical data using the median, quartiles, and potential outliers.

Here, I compare the age distributions of passengers who survived and those who did not.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Survived",
    y="Age",
    hue="Survived",
    palette="Set2",
    legend=False
)

plt.title("Age Distribution by Survival Status")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")

plt.tight_layout()
plt.show()
```

## 11. Box Plot: Fare by Passenger Class

This box plot compares ticket fares across the three passenger classes.

Box plots are particularly useful here because they show the median, spread, and potential extreme fare values within each class.

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Pclass",
    y="Fare",
    hue="Pclass",
    palette="Set3",
    legend=False
)

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

plt.tight_layout()
plt.show()
```

## 12. Scatter Plot: Age vs Fare

A scatter plot is useful for examining the relationship between two numerical variables.

Here, I plot passenger age against ticket fare and use survival status to distinguish observations.

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived",
    palette="coolwarm",
    alpha=0.7
)

plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.tight_layout()
plt.show()
```

## 13. Scatter Plot: Age vs Fare by Passenger Class

This visualization extends the previous scatter plot by using passenger class as the visual grouping variable.

It helps examine how age and fare are distributed across different passenger classes.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Pclass",
    palette="viridis",
    alpha=0.7
)

plt.title("Age vs Fare by Passenger Class")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.tight_layout()
plt.show()
```

## 14. Correlation Matrix

Correlation measures the strength and direction of a linear relationship between numerical variables.

I calculate the correlation matrix for selected numerical variables in the Titanic dataset.

A positive correlation indicates that two variables tend to increase together, while a negative correlation indicates that one tends to decrease as the other increases.

Correlation represents association and does not by itself establish causation.

```python
numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]

correlation_matrix = df[numerical_columns].corr()

correlation_matrix
```

## 15. Correlation Heatmap

A heatmap provides a visual representation of the correlation matrix.

The colors make it easier to identify stronger positive and negative relationships between numerical variables.

```python
plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    center=0
)

plt.title("Correlation Heatmap of Titanic Variables")

plt.tight_layout()
plt.show()
```

## 16. Custom Color Palette

Color palettes can improve the readability and visual consistency of charts.

Seaborn provides several built-in palettes such as `Set2`, `Set3`, `viridis`, and `coolwarm`. I demonstrate a custom palette below for categorical visualization.

```python
custom_palette = ["#4C72B0", "#DD8452", "#55A868"]

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Pclass",
    palette=custom_palette,
    legend=False
)

plt.title("Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.show()
```

## 17. Survival Rate by Passenger Class

Instead of displaying only the number of survivors, I calculate the survival rate for each passenger class.

Using percentages makes comparisons between groups easier because it accounts for differences in group size.

```python
survival_rate = (
    df.groupby("Pclass")["Survived"]
    .mean()
    .mul(100)
    .reset_index()
)

survival_rate.columns = ["Pclass", "SurvivalRate"]

survival_rate
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=survival_rate,
    x="Pclass",
    y="SurvivalRate",
    hue="Pclass",
    palette="viridis",
    legend=False
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")

plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

## 18. Survival Rate by Gender

This visualization focuses on survival rate rather than raw passenger counts.

Comparing percentages provides a clearer view of the proportion of passengers who survived within each gender group.

```python
gender_survival = (
    df.groupby("Sex")["Survived"]
    .mean()
    .mul(100)
    .reset_index()
)

gender_survival.columns = ["Sex", "SurvivalRate"]

gender_survival
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=gender_survival,
    x="Sex",
    y="SurvivalRate",
    hue="Sex",
    palette="Set2",
    legend=False
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")

plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```
